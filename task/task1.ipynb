```python
# 과제 요구사항: 
# 1. 사용자로부터 점수를 입력받습니다. 
# 2. 90점 이상: A 학점 
# 3. 80점 이상 90점 미만: B 학점 
# 4. 70점 이상 80점 미만: C 학점 
# 5. 60점 이상 70점 미만: D 학점 
# 6. 60점 미만: F 학점 
# 7. 0보다 작거나 100보다 큰 점수는 "잘못된 점수입니다." 출력
```

```python
score = 80
if score is None:
    score = int(input("점수를 입력하세요(0~100): "))

if score < 0 or score > 100:
    print("잘못된 점수입니다.")
elif score >= 90:
    print("A 학점")
elif score >= 80:
    print("B 학점")
elif score >= 70:
    print("C 학점")
elif score >= 60:
    print("D 학점")
else:
    print("F 학점")
```
