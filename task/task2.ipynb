```python
# while 문과 if문을 같이 사용하여 간단한 계산기 프로그램을 만들어 보세요.

# 사용자로부터 두 개의 숫자와 연산자(+, -, *, /)를 입력받아 계산 결과를 출력하는 프로그램을 작성하세요.
# while 문을 사용하여 계산을 반복하도록 구현하세요.
# 사용자가 'q'를 입력하면 프로그램을 종료하도록 하세요.
# 0으로 나누는 경우 "0으로 나눌 수 없습니다."를 출력하고 다시 입력을 받으세요.
# 잘못된 연산자가 입력되면 "잘못된 연산자입니다."를 출력하고 다시 입력을 받으세요.
```

```python
while True: 
    
    quit_input = input("계속하려면 Enter, 종료하려면 q를 입력하세요.")
    if quit_input == 'q':
        break

    numbers = [None, None]
    numbers[0] = int(input("첫 번째 숫자: "))
    operator = input("연산자를 입력하세요(+, -, *, /)")
    numbers[1] = int(input("두 번째 숫자: "))
    
    if operator == "+":
        print(numbers[0]+numbers[1])
    elif operator == "-":
        print(numbers[0]-numbers[1])
    elif operator == "*":
        print(numbers[0]*numbers[1])
    elif operator == "/":
        if numbers[1] == 0:
            print("0으로 나눌 수 없습니다.")
            continue
        print(numbers[0]/numbers[1])
    else:
        print("잘못된 연산자입니다.")
        continue
    
```
