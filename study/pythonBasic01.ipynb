### 리스트(list)

```python
fruits = ["apple", "banana", "cherry"]
print(fruits[0])
print(fruits[0:2])
```

```python
numbers = [3, 1, 4, 1, 5]
numbers.append(9)
print(numbers)
```

```python
numbers.sort()
print(numbers)
```

```python
numbers.remove(1)
print(numbers)
```

```python
removed_item = numbers.pop(2)   # pop : 특정 인덱스 위치의 요소 추출 및 삭제
print(numbers)
print(removed_item)
```

```python
numbers.reverse()
print(numbers)
```

```python
count = numbers.count(5)
print(count)
```

```python
new_numbers = numbers.copy()
print(new_numbers)
```

```python
numbers.extend([6,7,8])     # append는 값 1개, extend는 값 여러 개(리스트)
print(numbers)
```

```python
another_list = [10, 11, 12]
combined_list = numbers + another_list
print(combined_list)
```

#### 튜플(tuple)

```python
colors = ("red", "green", "blue")
print(colors[1])
print(colors[0:2])
```

```python
numbers = (1, 2, 3, 4, 5)
print(numbers.count(2)) 
print(numbers.index(3))     # index: 특정 값의 인덱스 찾기
```

```python
point = (3, 4)
print("x좌표 :", point[0])
print("y좌표 :", point[1])
```

```python
# 튜플 에러 : 수정할 수 없기 때문에 생기는 에러
my_tuple = (1, 2, 3, 4, 5)
my_tuple[0] = 10
my_tuple
```

```python
# 리스트는 값에 대한 수정 가능
my_list = [1, 2, 3, 4, 5]
my_list[0] = 10
print(my_list)
```

#### 딕셔너리

JSON 형태 -> 정보를 전달하기 위함

```python
person = {"name":"Bob", "age":30, "city":"New York"}
print(person["name"])
```

```python
# 값 수정
person["name"] = "Alice"
print(person["name"])
```

```python
# 키에 대응하는 값 반환
info = {"name":"Charlie", "age":22}
print(info.get("name"))
```

```python
# 모든 키를 반환
print(info.keys())
```

```python
# 모든 값을 반환
print(info.values())
```

```python
print(info.items())
```
