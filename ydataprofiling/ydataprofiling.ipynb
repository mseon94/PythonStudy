```python
import os
import pandas as pd
from ydata_profiling import ProfileReport

# 데이터 불러오기
file_path = 'dataset/diabetes.csv'
df = pd.read_csv(file_path)

print(df.info())
```

```python
# YData Profiling으로 데이터 프로파일링 보고서 생성
profile = ProfileReport(df, title="Diabetes Dataset Profiling Report", 
                        explorative=True)   # 자세하게 보여줘

# 보고서를 HTML 파일로 저장
output_file = 'report/diabetes_profiling_report.html'
profile.to_file(output_file)
print(f"프로파일링 보고서가 생성되었습니다: {output_file}")
```
